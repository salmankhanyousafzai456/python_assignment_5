## A traffic system calculates fines:

Speed limit = 60 km/h
* If speed ≤ 60 → No fine
* If 61–80 → Fine = 500
* If 81–100 → Fine = 2000
* If >100 → License suspended
#### BUT:
*If emergency vehicle → no fine
*If rainy weather AND speed > 60 → fine increases by 50%

```python
speed = int(input("Enter speed in km/h: "))
emergency = input("Is it an emergency vehicle? (yes/no): ").lower()
rainy = input("Is the weather rainy? (yes/no): ").lower()

if emergency == "yes":
    print("No fine")
else:
    if speed <= 60:
        print("No fine")

    elif speed <= 80:
        fine = 500

        if rainy == "yes":
            fine = fine * 1.5

        print("Fine =", fine)

    elif speed <= 100:
        fine = 2000

        if rainy == "yes":
            fine = fine * 1.5

        print("Fine =", fine)

    else:
        print("License suspended")
```

## A company filters candidates:
* Experience ≥ 5 years AND skill score ≥ 80 → shortlisted
* Experience 2–4 years AND skill score ≥ 70 → interview
* Freshers → only if test score ≥ 85
* If criminal record = True → reject immediately
* If multiple job switches (>3 in 2 years) → flag as unstable

```python
experience = int(input("Enter experience in years: "))
skill_score = int(input("Enter skill score: "))
test_score = int(input("Enter test score: "))
criminal_record = input("Criminal record? (yes/no): ").lower()
job_switches = int(input("Job switches in 2 years: "))

if criminal_record == "yes":
    print("Reject")
else:
    if job_switches > 3:
        print("Flag as unstable")

    if experience >= 5 and skill_score >= 80:
        print("Shortlisted")
    elif 2 <= experience <= 4 and skill_score >= 70:
        print("Interview")
    elif experience == 0 and test_score >= 85:
        print("Freshers eligible")
    else:
        print("Not selected")
```

## Write a program to decide passenger clearance status.
Take input:
* baggage_weight
* passport_valid (Yes/No)
* visa_status (Valid/Invalid)
* criminal_check (Clear/Flagged)
#### Rules:
* baggage > 30kg → extra charge
* passport invalid → reject
* visa invalid → reject
* criminal flagged → security hold
* otherwise → cleared for boarding

```python
baggage_weight = float(input("Enter baggage weight in kg: "))
passport_valid = input("Passport valid? (Yes/No): ").lower()
visa_status = input("Visa status (Valid/Invalid): ").lower()
criminal_check = input("Criminal check (Clear/Flagged): ").lower()

if passport_valid == "no":
    print("Reject")
elif visa_status == "invalid":
    print("Reject")
elif criminal_check == "flagged":
    print("Security Hold")
else:
    if baggage_weight > 30:
        print("Extra charge")
    print("Cleared for boarding")
```

## Write a program to decide insurance eligibility.
Take input:
* age
* smoker (Yes/No)
* existing_disease (Yes/No)
* annual_income
#### Rules:
* age < 18 → reject
* smoker + disease → high premium category
* income < 200000 → limited coverage
* age > 60 → special medical check required
* otherwise → normal approval

```python
age = int(input("Enter age: "))
smoker = input("Smoker? (Yes/No): ").lower()
disease = input("Existing disease? (Yes/No): ").lower()
income = float(input("Enter annual income: "))

if age < 18:
    print("Reject")
else:
    if smoker == "yes" and disease == "yes":
        print("High premium category")

    if income < 200000:
        print("Limited coverage")

    if age > 60:
        print("Special medical check required")

    if not (smoker == "yes" and disease == "yes") and income >= 200000 and age <= 60:
        print("Normal approval")
```

## Write a program to detect suspicious banking transactions.
Take input:
* transaction_amount
* location_match (Yes/No)
* time_of_transaction(Day/Night)
* previous_transactions_count
* device_verified (Yes/No)
#### Rules:
* amount > 100,000 → suspicious
* location mismatch + high amount → fraud alert
* night transaction + unverified device → high risk
* many transactions (>10 in 1 hour) → freeze account
* otherwise → normal transaction

```python
amount = float(input("Enter transaction amount: "))
location_match = input("Does location match? (Yes/No): ").lower()
time = input("Time of transaction (Day/Night): ").lower()
previous_transactions = int(input("Transactions in last 1 hour: "))
device_verified = input("Is device verified? (Yes/No): ").lower()

if previous_transactions > 10:
    print("Freeze account")
elif location_match == "no" and amount > 100000:
    print("Fraud alert")
elif time == "night" and device_verified == "no":
    print("High risk")
elif amount > 100000:
    print("Suspicious")
else:
    print("Normal transaction")
```

## Write a program to classify a person’s risk level in a police database.
Take input:
* criminal_record (Yes/No)
* number_of_cases
* travel_history_flagged (Yes/No)
* financial_activity_score
* last_activity_days
#### Rules:
* criminal_record = Yes AND cases > 3 → High Risk
* flagged travel + suspicious financial score → High Risk
* no activity for long time (>365 days) → Monitor
* clean record + normal activity → Low Risk
* mixed signals → Medium Risk

```python
criminal_record = input("Criminal record? (Yes/No): ").lower()
cases = int(input("Number of cases: "))
travel_flagged = input("Travel history flagged? (Yes/No): ").lower()
financial_score = int(input("Financial activity score: "))
last_activity = int(input("Days since last activity: "))

if criminal_record == "yes" and cases > 3:
    print("High Risk")

elif travel_flagged == "yes" and financial_score < 50:
    print("High Risk")

elif last_activity > 365:
    print("Monitor")

elif criminal_record == "no" and travel_flagged == "no" and financial_score >= 50:
    print("Low Risk")

else:
    print("Medium Risk")
```
